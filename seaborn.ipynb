```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
years=[2010,2011,2012,2013,2014,2015,2016,2017,2018,2019]
kohli=[250,400,550,700,900,1500,2200,2800,4500,8000]

sns.barplot(kohli)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Virat Kohli's Runs Over the Years")
plt.show()  
```

```python
sns.lineplot(kohli)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Virat Kohli's Runs Over the Years")
plt.show()
```

```python
sns.lineplot(years)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Virat Kohli's Runs Over the Years")
plt.show()
```

```python
sns.scatterplot(x=years, y=kohli)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Virat Kohli's Runs Over the Years")
plt.show()
```

```python
sns.boxplot(x=years, y=kohli)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Virat Kohli's Runs Over the Years")
plt.show()
```
