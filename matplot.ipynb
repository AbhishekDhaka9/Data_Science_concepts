```python
import numpy as np
list1=[1,3,2,5,9,7,4,12,8,6]
arr=np.array(list1)
print(arr)
```

```python
arr=np.sort(arr)
print(arr)
```

```python
import matplotlib.pyplot as plt
plt.plot(arr)
plt.show()
```

```python
years=[2010,2011,2012,2013,2014,2015,2016,2017,2018,2019]
kohli=[0,0,500,200,400,300,400,500,600,700]
sehwagn=[400,300,500,400,500,600,700,800, 900,1000]
yuvraj=[100,200,300,400,500,600,700,800,900,1000]

plt.plot(years,kohli,label='kohli',marker='o',color='red',linestyle='--')
plt.plot(years,sehwagn,label='sehwagn',marker='s',color='blue',linestyle='-.')
plt.plot(years,yuvraj,label='yuvraj',marker='^',color='green',linestyle=':')
plt.xlabel('Years')
plt.ylabel('Runs')
plt.title('Runs scored by players over years')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.bar(years,kohli,label='kohli',color='red')  
plt.bar(years,sehwagn,label='sehwagn',color='blue',alpha=0.5)
plt.bar(years,yuvraj,label='yuvraj',color='green',alpha=0.5)
plt.xlabel('Years') 
plt.ylabel('Runs')
plt.title('Runs scored by players over years')
plt.legend()
plt.grid()
plt.show()
```

```python
plt.pie(kohli, labels=years, autopct='%1.1f%%')
plt.title('Runs scored by Kohli over lifetime.')
plt.show()
```

```python
plt.barh(years,kohli,label='kohli',color='red')
plt.xlabel('Runs')
plt.ylabel('Years')
plt.title('Runs scored by Kohli over years')
plt.legend()
plt.style.use('ggplot')
plt.show()
```
