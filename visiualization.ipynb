```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data.csv")

plt.figure(figsize=(10,6))
for col in df.columns[1:]:
    plt.plot(df["Year"], df[col], label=col)

plt.title("Average Marks (2015-2025)")
plt.xlabel("Year")
plt.ylabel("Average Marks")
plt.legend()
plt.grid(True)
plt.show()
```
